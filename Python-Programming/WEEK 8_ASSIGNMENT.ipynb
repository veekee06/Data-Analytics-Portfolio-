```python
def check_number(num):
    if num > 0: 
        print("Positive") 
    elif num < 0: 
        print("Negative") 
    else: 
        print("Zero") 
```

```python
check_number(10) 
```

```python
check_number(-3) 
```

```python
check_number(0) 
```
